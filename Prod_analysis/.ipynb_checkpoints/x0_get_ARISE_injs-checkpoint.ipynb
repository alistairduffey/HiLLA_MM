```python
## this short script reads in log files for ARISE-SAI in both UKESM and CESM 
## to get injection magnitudes, and saves them as a dict to a .json 

import pandas as pd
import numpy as np
import json
## get injection magnitudes for final 20 years of each ARISE run from the log files
import config

#UKESM:
injs_UK = []
names = []
for run in config.UKESM_ARISE_run_IDs: 
    file = config.UKESM_arise_inj_logs_path+'FeedBack_stats_{}_2034.log'.format(run)
    log = pd.read_csv(file)
    log = log[log['Timestamp'] < config.assessment_period[1]+1]
    log = log[log['Timestamp'] > config.assessment_period[0]-1]
    injs_UK.append(np.mean(log[['30S(Tg)', '15S(Tg)', '15N(Tg)', '30N(Tg)']].sum(axis=1)))
    names.append(config.UKESM_ARISE_run_dict[run])
    
arise_injs_UK = dict(zip(names, injs_UK))

# CESM:

injs_CE = []
for run in config.CESM_ARISE_run_IDs: 
    file = config.CESM_arise_inj_logs_path + 'Controller_start_Default_{}.txt'.format(run)
    log = pd.read_csv(file, delimiter=' ')
    log = log[log['Timestamp'] < config.assessment_period[1]+1]
    log = log[log['Timestamp'] > config.assessment_period[0]-1]
    injs_CE.append(np.mean(log[['30S(Tg)', '15S(Tg)', '15N(Tg)', '30N(Tg)']].sum(axis=1)))

arise_injs_CE = dict(zip(config.CESM_ARISE_run_IDs, injs_CE))

print(arise_injs_UK, arise_injs_CE)

arise_injs = {'UKESM':arise_injs_UK, 'CESM':arise_injs_CE}

with open("../intermediate_outputs/arise_inj_mags.json", "w") as f:
    json.dump(arise_injs, f, indent=4)
```
