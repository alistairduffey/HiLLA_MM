```python
import s3fs

fs = s3fs.S3FileSystem(anon=True)  # Use anon=True for public buckets

def list_directories_recursive(fs, path, level=0, prefix=''):
    """Recursively list directories and create a tree structure."""
    try:
        items = fs.ls(path, detail=True)

        for item in items:
            # Skip if it's a zarr directory
            if item['name'].endswith('.zarr'):
                print(f"{prefix}{'└── ' if level > 0 else ''}{item['name'].split('/')[-1]}")
                continue

            # Print current item
            print(f"{prefix}{'└── ' if level > 0 else ''}{item['name'].split('/')[-1]}")

            # If it's a directory, recurse
            if item['type'] == 'directory':
                new_prefix = prefix + ('    ' if level == 0 else '    ')
                list_directories_recursive(fs, item['name'], level + 1, new_prefix)
    except Exception as e:
        print(f"Error accessing {path}: {str(e)}")

# Start the recursive listing from the bucket root
bucket_path = 's3://reflective-atmospheric-scientist-challenge/'
print("Directory Structure:")
list_directories_recursive(fs, bucket_path)
```
