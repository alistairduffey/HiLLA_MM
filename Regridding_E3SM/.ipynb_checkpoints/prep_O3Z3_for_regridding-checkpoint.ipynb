```python
import os
import glob
import pandas as pd
import numpy as np
import xarray as xr
from xmip.preprocessing import rename_cmip6
#import matplotlib.path as mpath
#import matplotlib.pyplot as plt
import json
import cftime
#import matplotlib
import warnings
#import matplotlib.pyplot as plt
#from Utils import get_ds_from_ceda
#from Utils import weighted_annual_resample
#import seaborn as sns
warnings.filterwarnings('ignore')
```

```python
def seasonal_time_slice(ds, start_yr_str='2050', end_yr_str='2069'):
    ds = ds.sel(time=slice(start_yr_str, end_yr_str)).groupby('time.month').mean(dim='time')
    return ds

def set_time_to_center_of_bounds(ds, time_bounds_name='time_bounds'):
    time_bounds = ds[time_bounds_name]
    time_midpoints = time_bounds.mean(dim=time_bounds.dims[-1])
    new_ds = ds.assign_coords(time=time_midpoints)
    return new_ds
    
path_archive = '/gws/ssde/j25b/cpom/aduffey/pp_archive_hills/'
```

```python
og_dir = path_archive+'E3SM_data/O3/'
new_dir = path_archive+'E3SM_data/O3_zm/'

for file in os.listdir(og_dir):
    print(file)
    ds = seasonal_time_slice(set_time_to_center_of_bounds(xr.open_dataset(og_dir+file), 'time_bnds'))['O3'].to_dataset(name='O3')
    ds.to_netcdf(new_dir+file)
    print('done')
```
